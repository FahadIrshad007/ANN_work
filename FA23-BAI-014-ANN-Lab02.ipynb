### Task 1: Vectors

```python
#TASK 01
ages= np.array([23,24,22,21,20,30])
print("Shape",np.shape(ages),"\nDimension:",np.ndim(ages),"\nType:",type(ages))
years_until_retirement = 65-np.array(ages)
print("Years until retirement:",years_until_retirement)
print("Average age:",np.average(ages))
```

### Task 3: Addition & Subtraction

```python
#TASK 03
term1=np.array([[85,89,80],[75,73,82],[90,95,99]])
term2=np.array([[82,79,70],[85,63,86],[92,93,90]])
Total=term1+term2
print("Total marks:\n",Total)
Improvement=term2-term1
print("Improvement:\n",Improvement)
highest=np.argmax(np.sum(Improvement,axis=1))
print("Student with highest total improvement:",highest+1)
```

### Task 4: Dot Product

```python
#TASK 04
prices=np.array([10,20,30,40])
quantities=np.array([10,25,38,49])
total_bill=np.dot(prices,quantities)
print("Total bill:",total_bill)
customers=np.array([[1,2,3,4],[5,6,7,8]])
total_bill_customers=np.dot(customers,prices)
print("Total bill for each customer:\n",total_bill_customers)
# The number of columns in customers must match the number of elements in prices.
# Here, customers has shape (2, 4) and prices has shape (4,).
# The 4s match, so the dot product can be performed.
```

### Task 7: Slicing

```python
#TASK 07
mat = np.arange(1, 37).reshape(6, 6)
print(mat)
print(mat[:3, :3])
print(mat[::2, ::2])
print(mat[-2:, -2:])
```

### Task 8: Random Matrices

```python
#TASK 08
np.random.seed(67)
random_decimal = np.random.rand(3, 3)
print("Random decimal matrix:\n", random_decimal)
random_integer = np.random.randint(10, 100, (3, 3))
print("Random integer matrix:\n", random_integer)
random_normal = np.random.randn(4, 4)
print("Random normal matrix:\n", random_normal)
# Both outputs are identical because resetting the seed to the same value
# makes NumPy generate the same sequence of random numbers.
```

### Task 9: Transpose

```python
#TASK 09
employees = np.array([[85,89,80,79,90],[75,73,82,99,98],[90,95,99,67,64]])
print("Transposed matrix:\n", employees.T,"\nNew shape:",np.shape(employees.T))
print("Original matrix:\n", employees.T.T)
employees_2=np.array([[85,89],[75,73],[90,95],[99,67]])
employees_3=np.array([[85,89],[75,73],[90,95]])
print("Dot product:\n",np.dot(employees_2,employees_3.T))
```
